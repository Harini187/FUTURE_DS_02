```python
import pandas as pd
import numpy as np

# 1. Load raw data
df = pd.read_csv("extracted_data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# 2. Convert empty spaces to NaN variables
df['TotalCharges'] = df['TotalCharges'].replace(" ", np.nan)

# 3. Transform format type into analytical float numbers
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])

# 4. FIX: Direct overwrite method to completely bypass pandas warnings
total_charges_median = df['TotalCharges'].median()
df['TotalCharges'] = df['TotalCharges'].fillna(total_charges_median)

# 5. Map target string values to execution binary metrics
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

print("Step 1 clean code executed successfully!")
```

```python
# =========================================================================
# STEP 2: COHORT ANALYSIS & SEGMENTATION (BACKEND CALCULATIONS)
# =========================================================================

# 1. Segment customer tenure into 4 strategic lifecycle cohorts
def lifecycle_bucket(months):
    if months <= 6:
        # Target: Immediate drop-offs
        return '0-6 Months (New Users)'
    elif months <= 12:
        # Target: Early lifecycle retention risk
        return '6-12 Months (Early Lifecycle)'
    elif months <= 24:
        # Target: Mid-term stabilization window
        return '1-2 Years (Mid Tenure)'
    else:
        # Target: Fully retained core customer base
        return '2+ Years (Loyal Cohort)'

# Create a new analytical column using the mapping logic
df['Lifecycle_Stage'] = df['tenure'].apply(lifecycle_bucket)

# 2. Extract metrics to answer the primary business goals
# Metric A: Churn performance over subscription lifecycle timelines
cohort_summary = df.groupby('Lifecycle_Stage')['Churn_Numeric'].mean().reset_index()
cohort_summary.rename(columns={'Churn_Numeric': 'Churn_Rate'}, inplace=True)

# Metric B: High-risk operational analysis by subscriber contract frameworks
contract_summary = df.groupby('Contract')['Churn_Numeric'].mean().reset_index()
contract_summary.rename(columns={'Churn_Numeric': 'Churn_Rate'}, inplace=True)

# Metric C: Product evaluation based on technical support assistance availability
support_summary = df.groupby('TechSupport')['Churn_Numeric'].mean().reset_index()
support_summary.rename(columns={'Churn_Numeric': 'Churn_Rate'}, inplace=True)

print("Step 2 completed successfully!.")
```

```python
# =========================================================================
# STEP 3: EXPORT TO MULTI-SHEET EXCEL WORKBOOK
# =========================================================================
import pandas as pd

excel_filename = 'final_churn_analytics.xlsx'

# Sync and write all summaries into individual clean sheets
with pd.ExcelWriter(excel_filename) as writer:
    df.to_excel(writer, sheet_name='Master_Data', index=False)
    cohort_summary.to_excel(writer, sheet_name='Cohort_Overview', index=False)
    contract_summary.to_excel(writer, sheet_name='Contract_Overview', index=False)
    support_summary.to_excel(writer, sheet_name='Support_Overview', index=False)

print(f"Success! '{excel_filename}' is ready. Download it now.")
```
